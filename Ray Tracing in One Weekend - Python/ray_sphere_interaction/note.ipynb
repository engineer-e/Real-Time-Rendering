# 5. Adding a Sphere

---

## 5.1 Ray–Sphere Intersection (Step by Step)

We want to find where a ray hits a sphere.

### Given:

Ray equation:

$$
P(t)=Q+t\vec d
$$

where:

* $Q$ = ray origin
* $\vec d$ = ray direction
* $t$ = parameter

Sphere equation:

$$
(x-C_x)^2+(y-C_y)^2+(z-C_z)^2=r^2
$$

Vector form:

$$
(P-C)\cdot(P-C)=r^2
$$

where:

* $C$ = sphere center
* $r$ = radius

---

## Step-by-step derivation

| Step | Equation                                                                                | Explanation                    |
| ---: | --------------------------------------------------------------------------------------- | ------------------------------ |
|    1 | $$ (P-C)\cdot(P-C)=r^2 $$                                                               | Sphere equation                |
|    2 | $$ P(t)=Q+t\vec d $$                                                                    | Substitute ray equation        |
|    3 | $$ (Q+t\vec d-C)\cdot(Q+t\vec d-C)=r^2 $$                                               | Replace $P$                    |
|    4 | $$ ((Q-C)+t\vec d)\cdot((Q-C)+t\vec d)=r^2 $$                                           | Group constants                |
|    5 | $$ (Q-C)\cdot(Q-C) + (Q-C)\cdot t\vec d + t\vec d\cdot(Q-C)+t\vec d\cdot t\vec d=r^2 $$ | Expand dot product             |
|    6 | $$ (Q-C)\cdot(Q-C)+2t\vec d\cdot(Q-C)+t^2\vec d\cdot\vec d=r^2 $$                       | Combine middle terms           |
|    7 | $$ t^2(\vec d\cdot\vec d)+2t(\vec d\cdot(Q-C))+(Q-C)\cdot(Q-C)-r^2=0 $$                 | Rearranged into quadratic form |

---

# Compare with standard quadratic

Standard quadratic:

$$
at^2+bt+c=0
$$

Match terms:

| Coefficient | Value                     |
| ----------- | ------------------------- |
| $a$         | $$ \vec d\cdot\vec d $$   |
| $b$         | $$ 2\vec d\cdot(Q-C) $$   |
| $c$         | $$ (Q-C)\cdot(Q-C)-r^2 $$ |

---

# Solve for $t$

Quadratic formula:

$$
t=\frac{-b\pm\sqrt{b^2-4ac}}{2a}
$$

Substitute:

$$
t=\frac{-2\vec d\cdot(Q-C)\pm\sqrt{(2\vec d\cdot(Q-C))^2-4(\vec d\cdot\vec d)((Q-C)\cdot(Q-C)-r^2)}}{2(\vec d\cdot\vec d)}
$$

Simplified:

$$
t=\frac{-\vec d\cdot(Q-C)\pm\sqrt{(\vec d\cdot(Q-C))^2-(\vec d\cdot\vec d)((Q-C)\cdot(Q-C)-r^2)}}{\vec d\cdot\vec d}
$$

---

# Discriminant

The important part:

$$
\Delta=b^2-4ac
$$

| Condition      | Meaning   | Geometry                     |
| -------------- | --------- | ---------------------------- |
| $$\Delta > 0$$ | Two roots | Ray enters and exits sphere  |
| $$\Delta = 0$$ | One root  | Ray touches sphere (tangent) |
| $$\Delta < 0$$ | No root   | Ray misses sphere            |

---

# Geometric meaning

Think of it like this:

```text
Δ > 0   →   ----(sphere)----   (2 hits)
Δ = 0   →   ----o              (touch)
Δ < 0   →      (sphere)   ---- (miss)
```

---

# Optimization trick (used in ray tracers)

Instead of:

$$
b=2\vec d\cdot(Q-C)
$$

Use:

$$
h=\vec d\cdot(C-Q)
$$

Then:

$$
a=\vec d\cdot\vec d
$$

$$
c=(C-Q)\cdot(C-Q)-r^2
$$

Solve:

$$
t=\frac{h\pm\sqrt{h^2-ac}}{a}
$$

This avoids multiplying by 2.

---

This is the exact mathematical foundation used in books like **Ray Tracing in One Weekend**.
