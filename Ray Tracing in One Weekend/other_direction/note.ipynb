## 1. Triangles (The Real World Geometry Format)

### What it is:

Almost everything in 3D graphics is made of **triangles**.

Not spheres. Not quads. Not implicit shapes.

Just triangles.

---

### Why triangles matter:

* Any complex object (human, car, tree) → triangle mesh
* GPUs are optimized for triangles
* Industry standard format

---

### The challenge (important):

The book hints:

> “model I/O is the worst”

Because you must handle:

* OBJ / FBX / GLTF parsing
* huge mesh files (millions of triangles)
* memory + acceleration structures (BVH becomes essential here)

---

### Key idea:

Once you support triangles:

> You can render ANY 3D model in existence.

---

## 2. Parallelism (Render Faster Using Many Cores)

### Core idea:

Instead of one CPU core:

> use many cores or machines

Each one renders:

* same image
* different random seeds

Then you average results.

---

### Why this works:

Ray tracing is Monte Carlo based:

* noise reduces as you average more samples
* randomness is actually useful

So:

> more independent renders → cleaner image

---

### Hierarchical averaging idea:

They suggest a tree-like structure:

```
N renders
→ pairwise average
→ N/2 images
→ repeat
→ final image
```

This is powerful because:

* scalable to thousands of cores
* minimal synchronization
* used in render farms

---

### Real-world connection:

This is exactly how:

* Pixar
* Blender render farms
* VFX studios

render images overnight.

---

## 3. Shadow Rays (Direct Lighting Control)

### What it is:

A **shadow ray** is a special ray:

> “Is this point visible to the light?”

---

### How it works:

For a surface point:

1. Find light source
2. Shoot ray toward light
3. If blocked → shadow
4. If not blocked → light contributes

---

### Why it matters:

This gives:

* sharp shadows
* soft shadows (if area lights are used)
* much more efficient lighting

---

### Key insight:

Instead of waiting for random bounces to hit light:

> you directly ask the light

This is called:

> **Next Event Estimation (NEE)** in advanced rendering
