# PPM Image Rendering (Gradient Generator)

## Purpose

This program creates a simple image in **PPM (Portable Pixmap)** format.

Purpose:

* Understand pixels
* Learn RGB color math
* Generate images using formulas
* Build graphics foundations

---

# 1. Image Size

```python
image_width = 256
image_height = 256
```

## Formula

$$
TotalPixels = Width \times Height
$$

Substitute:

$$
TotalPixels = 256 \times 256
$$

$$
= 65536
$$

### Description

Width = horizontal pixels
Height = vertical pixels

### Purpose

Defines image resolution.

---

# 2. PPM Header

```python
f.write(f"P3\n{image_width} {image_height}\n255\n")
```

PPM format:

```text
P3
Width Height
MaxColor
PixelData
```

---

## Formula

ASCII format:

$$
P3 = Plain\ Text\ RGB
$$

Image dimension:

$$
(Width, Height)
$$

Color range:

$$
0 \le Color \le 255
$$

---

### Purpose

The header tells the image viewer:

* file type
* size
* maximum color value

---

# 3. Pixel Coordinates

```python
for j in range(image_height):
    for i in range(image_width):
```

## Formula

Each pixel:

$$
(i,j)
$$

Where:

$$
0 \le i < Width
$$

$$
0 \le j < Height
$$

---

### Purpose

Traverse every pixel.

Total loop operations:

$$
Width \times Height
$$

---

# 4. Normalize Color Values

```python
r = i / (image_width - 1)
g = j / (image_height - 1)
b = 0.0
```

---

## Red Formula

$$
r = \frac{i}{Width-1}
$$

Range:

$$
0 \le r \le 1
$$

Purpose:

Horizontal gradient.

---

## Green Formula

$$
g = \frac{j}{Height-1}
$$

Range:

$$
0 \le g \le 1
$$

Purpose:

Vertical gradient.

---

## Blue Formula

$$
b = 0
$$

Purpose:

No blue.

---

Why divide by `(Width-1)`?

First pixel:

$$
\frac{0}{255}=0
$$

Last pixel:

$$
\frac{255}{255}=1
$$

This normalizes perfectly.

---

# 5. Convert Float to RGB Integer

```python
ir = int(255.999 * r)
ig = int(255.999 * g)
ib = int(255.999 * b)
```

Formula:

$$
IntegerColor = 255.999 \times FloatColor
$$

Range:

$$
0 \le IntegerColor \le 255
$$

Example:

If:

$$
r = 0.5
$$

Then:

$$
ir = 255.999 \times 0.5
$$

$$
ir = 127.999
$$

After integer conversion:

$$
ir = 127
$$

---

### Purpose

PPM requires integer RGB values.

---

# 6. Writing Pixel Data

```python
f.write(f"{ir} {ig} {ib}\n")
```

Formula:

$$
Pixel = (R,G,B)
$$

Example:

$$
(255,0,0)
$$

means pure red.

---

# Final Color Function

Overall color formula:

$$
Color(x,y)=
\left(
\frac{x}{W-1},
\frac{y}{H-1},
0
\right)
$$

Converted to RGB:

$$
Pixel(x,y)=
\left(
255\frac{x}{W-1},
255\frac{y}{H-1},
0
\right)
$$

---

# Output Behavior

Left side:

$$
R=0
$$

Right side:

$$
R=255
$$

Bottom:

$$
G=0
$$

Top:

$$
G=255
$$

Blue:

$$
B=0
$$

So final image:

$$
Black \rightarrow Red \rightarrow Yellow
$$

because:

$$
Yellow = Red + Green
$$
