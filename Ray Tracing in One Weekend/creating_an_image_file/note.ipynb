## 2.2 Creating an Image File


```python
# Image
image_width = 256
image_height = 256



# Render
with open("image.ppm", "w") as f:
    f.write(f"P3\n{image_width} {image_height}\n255\n")
    for j in range(image_height):
        for i in range(image_width):
            r = i / (image_width - 1)
            g = j / (image_height - 1)
            b = 0.0


            ir = int(255.999 * r)
            ig = int(255.999 * g)
            ib = int(255.999 * b)

            f.write(f"{ir} {ig} {ib}\n")   
```

# PPM Gradient Renderer — Table Format

| Step                      | Code                                                              | Formula                                                    | Description                                               | Purpose                                       |
| ------------------------- | ----------------------------------------------------------------- | ---------------------------------------------------------- | --------------------------------------------------------- | --------------------------------------------- |
| **1. Image Size**         | `image_width = 256` `image_height = 256`                          | $$\text{Pixels} = W \times H$$ $$=256 \times 256 = 65536$$ | Defines total pixels in the image.                        | Sets image resolution.                        |
| **2. PPM Header**         | `f.write(f"P3\n{image_width} {image_height}\n255\n")`             | $$P3$$ $$C_{max}=255$$                                     | `P3` means plain-text RGB. `255` is max color intensity.  | Tells the file format, size, and color range. |
| **3. Pixel Loop**         | `for j in range(image_height):` `for i in range(image_width):`    | $$ (i,j) $$ $$0 \le i < W$$ $$0 \le j < H$$                | Visits every pixel coordinate.                            | Traverses the image grid.                     |
| **4. Red Value**          | `r = i / (image_width - 1)`                                       | $$r=\frac{i}{W-1}$$                                        | Converts x-position into normalized red value.            | Creates horizontal red gradient.              |
| **5. Green Value**        | `g = j / (image_height - 1)`                                      | $$g=\frac{j}{H-1}$$                                        | Converts y-position into normalized green value.          | Creates vertical green gradient.              |
| **6. Blue Value**         | `b = 0.0`                                                         | $$b=0$$                                                    | Blue channel fixed to zero.                               | Removes blue from image.                      |
| **7. Integer Conversion** | `ir = int(255.999*r)` `ig = int(255.999*g)` `ib = int(255.999*b)` | $$I=255.999x$$                                             | Converts normalized float values into integer RGB values. | PPM requires integer colors.                  |
| **8. Write Pixel**        | `f.write(f"{ir} {ig} {ib}\n")`                                    | $$Pixel=(R,G,B)$$                                          | Stores each pixel into the file.                          | Builds final image data.                      |

---

# Coordinate → Color Mapping

| Position     | Formula                    | RGB Output    | Color  |
| ------------ | -------------------------- | ------------- | ------ |
| Bottom-left  | $$\left(0,0,0\right)$$     | `(0,0,0)`     | Black  |
| Bottom-right | $$\left(255,0,0\right)$$   | `(255,0,0)`   | Red    |
| Top-left     | $$\left(0,255,0\right)$$   | `(0,255,0)`   | Green  |
| Top-right    | $$\left(255,255,0\right)$$ | `(255,255,0)` | Yellow |

---

# Full Color Equation

| Type             | Formula                                                     |
| ---------------- | ----------------------------------------------------------- |
| Normalized color | $$C(x,y)=\left(\frac{x}{W-1},\frac{y}{H-1},0\right)$$       |
| Integer RGB      | $$P(x,y)=\left(255\frac{x}{W-1},255\frac{y}{H-1},0\right)$$ |

---

# Rendering Pipeline

| Stage      | Formula                                        |
| ---------- | ---------------------------------------------- |
| Coordinate | $$(x,y)$$                                      |
| Normalize  | $$\left(\frac{x}{W-1},\frac{y}{H-1},0\right)$$ |
| Scale      | $$255 \cdot color$$                            |
| Integer    | $$int(255.999 \cdot color)$$                   |
| Write      | $$(R,G,B)$$                                    |

---

This is the basic graphics pipeline:

$$
\text{Coordinate} \rightarrow \text{Color} \rightarrow \text{Pixel} \rightarrow \text Image
$$
