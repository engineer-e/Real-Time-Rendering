## 2.1 The PPM Image Format


- To Start rendering, you need image.
- To Write the pixel data to image using '*.ppm' file

- File name is "image.ppm".
```ppm
P3
3 2
255
255 0 0 0 255 0 0 0 255
255 255 0 255 255 255 0 0 0
```

- Add the above code to "image.ppm" file, then use below python code to visualize it.

````python
from PIL import Image
import matplotlib.pyplot as plt

img = Image.open("image.ppm")

scale = 10
zoomed = img.resize(
    (img.width * scale, img.height * scale),
    Image.NEAREST
)

plt.imshow(zoomed)
plt.axis("off")
plt.show()
````
