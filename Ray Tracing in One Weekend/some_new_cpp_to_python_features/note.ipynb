## 6.6 Some New C++ Features 

In Python, the same ideas from `shared_ptr`, `make_shared`, and `std::vector` exist—but they are *built into the language*, so you don’t explicitly manage them.

Let’s map each C++ feature into Python with **learning notes + purpose + “formula-style” understanding**.

---

# 1. Core Idea Mapping (C++ → Python Mental Model)

| Concept            | C++                   | Python Equivalent       | Key Idea                                      |
| ------------------ | --------------------- | ----------------------- | --------------------------------------------- |
| Smart pointer      | `std::shared_ptr<T>`  | normal object reference | Python uses automatic reference counting + GC |
| Object creation    | `make_shared<T>(...)` | `T(...)`                | direct constructor call                       |
| Dynamic array      | `std::vector<T>`      | `list`                  | dynamic resizing array                        |
| Namespace shortcut | `using std::vector`   | not needed              | Python imports directly                       |

---

# 2. Shared Ownership (C++ `shared_ptr`) in Python

## 💡 Concept

In C++:

$$
\text{shared-ptr} = \text{reference-counted ownership pointer}
$$

In Python:

$$
\text{object lifetime} = \text{reference counting + garbage collector}
$$

---

## 📌 Python Equivalent Example

```python
class Sphere:
    def __init__(self, center, radius):
        self.center = center
        self.radius = radius


sphere_ptr = Sphere((0, 0, 0), 1.0)
another_ref = sphere_ptr   # shared reference
```

---

## 📊 Explanation Table

| Step             | Formula / Idea           | Purpose           | What happens internally       |
| ---------------- | ------------------------ | ----------------- | ----------------------------- |
| Create object    | $o = \text{Sphere}(...)$ | allocate object   | Python creates object in heap |
| Assign reference | $r_2 = r_1$              | share ownership   | reference count increases     |
| Delete reference | `del r1`                 | release one owner | reference count decreases     |
| Auto cleanup     | $refcount = 0$           | free memory       | GC destroys object            |

---

## 🔥 Key Insight

C++:

```cpp
auto s = make_shared<Sphere>(...);
```

Python:

```python
s = Sphere(...)
```

👉 Python always behaves like a `shared_ptr` unless you explicitly break references.

---

# 3. `make_shared` vs Python Constructor

## 📌 Formula View

C++:

$$
\text{make-shared}(T, args) \rightarrow \text{shared-ptr<T>}
$$

Python:

$$
T(args) \rightarrow \text{object reference}
$$

---

## 📊 Table

| Operation       | C++                    | Python           | Purpose                      |
| --------------- | ---------------------- | ---------------- | ---------------------------- |
| allocate object | `make_shared<T>(args)` | `T(args)`        | create object in heap        |
| return pointer  | `shared_ptr<T>`        | object reference | automatic memory-safe handle |

---

# 4. `std::vector` → Python `list`

## 💡 Concept

C++:

$$
\text{vector} = \text{contiguous dynamic array}
$$

Python:

$$
\text{list} = \text{dynamic array of references}
$$

---

## 📌 Python Example

```python
objects = []

objects.append(Sphere((0,0,0), 1.0))
objects.append(Sphere((1,0,0), 0.5))
```

---

## 📊 Table

| Operation        | Formula                   | Purpose             |
| ---------------- | ------------------------- | ------------------- |
| create container | $V = []$                  | empty dynamic list  |
| add element      | $V \leftarrow V \cup {x}$ | append object       |
| access element   | $V[i]$                    | retrieve reference  |
| iterate          | $\forall x \in V$         | process all objects |

---

## 🔥 Key Insight

C++:

```cpp
std::vector<std::shared_ptr<hittable>> objects;
```

Python:

```python
objects = []   # list of objects
```

No pointer syntax needed.

# 5. hittable_list Equivalent in Python (Ray Tracing Style)

## 📌 Python Version

```python
class HittableList:
    def __init__(self):
        self.objects = []

    def add(self, obj):
        self.objects.append(obj)

    def hit(self, ray):
        hit_anything = False
        closest_so_far = float('inf')
        hit_record = None

        for obj in self.objects:
            hit = obj.hit(ray, 0.001, closest_so_far)
            if hit:
                hit_anything = True
                closest_so_far = hit.t
                hit_record = hit

        return hit_anything, hit_record
```

---

## 📊 Step-by-step Meaning

| Step               | Mathematical Idea      | Purpose               |
| ------------------ | ---------------------- | --------------------- |
| initialize list    | $L = \emptyset$        | store scene objects   |
| iterate objects    | $\forall o \in L$      | test ray intersection |
| update closest hit | $t_{min} = \min(t)$    | find nearest surface  |
| return result      | $(\text{hit}, record)$ | shading input         |

# 6. Memory Model Comparison

| Feature                  | C++ `shared_ptr`              | Python             |
| ------------------------ | ----------------------------- | ------------------ |
| ownership tracking       | manual but automatic refcount | fully automatic    |
| deletion                 | when count = 0                | GC + refcount      |
| circular reference issue | not automatic                 | handled by GC      |
| performance              | faster, explicit              | slower, abstracted |

---

# 7. Big Picture Intuition

Think of it like this:

### C++ world:

You say:

> “This object is shared. Track owners carefully.”

### Python world:

You say:

> “Just create objects and use them. Runtime will manage sharing.”

---

# 8. Final Mental Formula

### C++:

$$
\text{Scene} = { \text{shared-ptr}(object_i) }_{i=1}^{n}
$$

### Python:

$$
\text{Scene} = [object_1, object_2, \dots, object_n]
$$

Same math structure — different memory philosophy.
