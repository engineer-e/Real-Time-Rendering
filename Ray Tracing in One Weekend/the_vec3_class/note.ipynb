# `vec3` Class in Python

A `vec3` is simply a **3-dimensional vector**.

It stores:

| Meaning                   | Example   |
| ------------------------- | --------- |
| Position (point in space) | $(x,y,z)$ |
| Direction                 | $(1,0,0)$ |
| Color                     | $(R,G,B)$ |

So:

$$
\vec{v} = (x,y,z)
$$

Think of it like:

* **Point** → where something is
* **Direction** → where something moves
* **Color** → how bright it is

Same container, different meaning.

# Python Code (`vec3.py`)

```python
import math


class Vec3:
    def __init__(self, x=0, y=0, z=0):
        self.e = [x, y, z]

    # Accessors
    @property
    def x(self):
        return self.e[0]

    @property
    def y(self):
        return self.e[1]

    @property
    def z(self):
        return self.e[2]

    # Negative vector
    def __neg__(self):
        return Vec3(-self.e[0], -self.e[1], -self.e[2])

    # Index access
    def __getitem__(self, i):
        return self.e[i]

    def __setitem__(self, i, value):
        self.e[i] = value

    # Addition
    def __add__(self, other):
        return Vec3(
            self.e[0] + other.e[0],
            self.e[1] + other.e[1],
            self.e[2] + other.e[2]
        )

    # Subtraction
    def __sub__(self, other):
        return Vec3(
            self.e[0] - other.e[0],
            self.e[1] - other.e[1],
            self.e[2] - other.e[2]
        )

    # Multiplication
    def __mul__(self, t):
        if isinstance(t, Vec3):
            return Vec3(
                self.e[0] * t.e[0],
                self.e[1] * t.e[1],
                self.e[2] * t.e[2]
            )
        return Vec3(
            self.e[0] * t,
            self.e[1] * t,
            self.e[2] * t
        )

    # Division
    def __truediv__(self, t):
        return self * (1 / t)

    # Length squared
    def length_squared(self):
        return (
            self.e[0]**2 +
            self.e[1]**2 +
            self.e[2]**2
        )

    # Length
    def length(self):
        return math.sqrt(self.length_squared())

    def __str__(self):
        return f"{self.e[0]} {self.e[1]} {self.e[2]}"


# Dot product
def dot(u, v):
    return u.e[0]*v.e[0] + u.e[1]*v.e[1] + u.e[2]*v.e[2]


# Cross product
def cross(u, v):
    return Vec3(
        u.e[1]*v.e[2] - u.e[2]*v.e[1],
        u.e[2]*v.e[0] - u.e[0]*v.e[2],
        u.e[0]*v.e[1] - u.e[1]*v.e[0]
    )


# Unit vector
def unit_vector(v):
    return v / v.length()
```

# `vec3` Mathematical Meaning Table

| Operation                   | Formula                                                              | Description                         | Purpose                             |
| --------------------------- | -------------------------------------------------------------------- | ----------------------------------- | ----------------------------------- |
| **Vector Form**             | $\vec{v} = (x,y,z)$                                                  | A 3D quantity with three components | Stores position, direction, color   |
| **Addition**                | $\vec{u}+\vec{v}=(u_x+v_x,;u_y+v_y,;u_z+v_z)$                        | Combine two vectors                 | Move points, combine forces/colors  |
| **Subtraction**             | $\vec{u}-\vec{v}=(u_x-v_x,;u_y-v_y,;u_z-v_z)$                        | Difference between vectors          | Direction from one point to another |
| **Scalar Multiplication**   | $t\vec{v}=(tx,;ty,;tz)$                                              | Scale vector by number              | Stretch or shrink                   |
| **Scalar Division**         | $\vec{v}/t=(x/t,;y/t,;z/t)$                                          | Reduce vector size                  | Normalize or scale down             |
| **Negation**                | $-\vec{v}=(-x,;-y,;-z)$                                              | Reverse direction                   | Opposite movement                   |
| **Length Squared**          | $\|\vec{v}\|^2=x^2+y^2+z^2$                                            | Magnitude without square root       | Faster comparisons                  |
| **Length (Magnitude)**      | $\|\vec{v}\|=\sqrt{x^2+y^2+z^2}$                                       | Total size of vector                | Distance, speed                     |
| **Dot Product**             | $\vec{u}\cdot\vec{v}=u_xv_x+u_yv_y+u_zv_z$                           | Measures alignment                  | Angle, lighting, projection         |
| **Dot Angle Relation**      | $\vec{u}\cdot\vec{v}=\|\vec{u}\|\|\vec{v}\|\cos\theta$                   | Connects dot product to angle       | Find angle between vectors          |
| **Cross Product**           | $\vec{u}\times\vec{v}=(u_yv_z-u_zv_y,;u_zv_x-u_xv_z,;u_xv_y-u_yv_x)$  | Produces perpendicular vector       | Surface normals, orientation        |
| |  $$ \vec{u}\times\vec{v}=\begin{bmatrix} i & j & k \\ u_x & u_y & u_z \\ v_x & v_y & v_z \end{bmatrix}$$ | | |
| **Unit Vector**             | $\hat{v}=\vec{v}/\|\vec{v}\|$                                          | Keeps direction, makes length = 1   | Normalize direction                 |
| **Distance Between Points** | $d=\|\vec{A}-\vec{B}\|$                                                | Distance between two positions      | Collision, ray hit tests            |
| **Ray Equation**            | $P(t)=A+tB$                                                          | Point moving along a ray            | Ray tracing foundation              |

## Quick Geometric Meaning

| Formula                 | Meaning               |
| ----------------------- | --------------------- |
| $\vec{A}+\vec{B}$       | Go along A, then B    |

| $\vec{A}\cdot\vec{B}=0$ | Perpendicular         |
| $\vec{A}\times\vec{B}$  | Normal to both        |

# Why Ray Tracing Needs This

| Thing           | vec3 use |
| --------------- | -------- |
| Camera position | point3   |
| Ray direction   | vec3     |
| Object position | point3   |
| Surface normal  | vec3     |
| Pixel color     | color    |

Ray equation:

$$
P(t)=A+tB
$$

Where:

| Symbol | Meaning            |
| ------ | ------------------ |
| $A$    | origin (`point3`)  |
| $B$    | direction (`vec3`) |
| $t$    | distance           |

Without `vec3`, ray tracing cannot exist.

It is the **foundation**.


````python
import math


class Vec3:
    def __init__(self, x=0, y=0, z=0):
        self.e = [x, y, z]

    # Accessors
    @property
    def x(self):
        return self.e[0]

    @property
    def y(self):
        return self.e[1]

    @property
    def z(self):
        return self.e[2]

    # Negative vector
    def __neg__(self):
        return Vec3(-self.e[0], -self.e[1], -self.e[2])

    # Index access
    def __getitem__(self, i):
        return self.e[i]

    def __setitem__(self, i, value):
        self.e[i] = value

    # Addition
    def __add__(self, other):
        return Vec3(
            self.e[0] + other.e[0],
            self.e[1] + other.e[1],
            self.e[2] + other.e[2]
        )

    # Subtraction
    def __sub__(self, other):
        return Vec3(
            self.e[0] - other.e[0],
            self.e[1] - other.e[1],
            self.e[2] - other.e[2]
        )

    # Multiplication
    def __mul__(self, t):
        if isinstance(t, Vec3):
            return Vec3(
                self.e[0] * t.e[0],
                self.e[1] * t.e[1],
                self.e[2] * t.e[2]
            )
        return Vec3(
            self.e[0] * t,
            self.e[1] * t,
            self.e[2] * t
        )

    # Division
    def __truediv__(self, t):
        return self * (1 / t)

    # Length squared
    def length_squared(self):
        return (
            self.e[0]**2 +
            self.e[1]**2 +
            self.e[2]**2
        )

    # Length
    def length(self):
        return math.sqrt(self.length_squared())

    def __str__(self):
        return f"{self.e[0]} {self.e[1]} {self.e[2]}"


# Dot product
def dot(u, v):
    return u.e[0]*v.e[0] + u.e[1]*v.e[1] + u.e[2]*v.e[2]


# Cross product
def cross(u, v):
    return Vec3(
        u.e[1]*v.e[2] - u.e[2]*v.e[1],
        u.e[2]*v.e[0] - u.e[0]*v.e[2],
        u.e[0]*v.e[1] - u.e[1]*v.e[0]
    )


# Unit vector
def unit_vector(v):
    return v / v.length()
````

````python
a = Vec3(1, 2, 3)
b = Vec3(4, 5, 6)

print("Add:", a + b)
print("Sub:", a - b)
print("Dot:", dot(a, b))
print("Cross:", cross(a, b))
print("Length:", a.length())
print("Unit:", unit_vector(a))
````
