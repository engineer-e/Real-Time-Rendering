# Ray Class in Python

```python
# ray.py
from vec3 import Vec3

Point3 = Vec3   # point is just a vector

class Ray:
    def __init__(self, origin, direction):
        self.orig = origin
        self.dir = direction

    def origin(self):
        return self.orig

    def direction(self):
        return self.dir

    def at(self, t):
        return self.orig + t * self.dir
```

---

# Learning Notes

| Topic                 | Explanation                                                                        |
| --------------------- | ---------------------------------------------------------------------------------- |
| **What is a Ray?**    | A ray is like an arrow starting from one point and going in one direction forever. |
| **Mathematical Form** | $$P(t) = A + tb$$                                                                  |
| **Origin ($A$)**      | Starting point of the ray.                                                         |
| **Direction ($b$)**   | Which way the ray moves.                                                           |
| **Parameter ($t$)**   | Controls how far we travel along the ray.                                          |
| **Position ($P(t)$)** | Gives the exact point at time/step $t$.                                            |
| **Negative $t$**      | Goes behind the origin.                                                            |
| **Positive $t$**      | Goes forward (real ray tracing uses this).                                         |

---

# Formula Breakdown

| Formula                                   | Meaning             |
| ----------------------------------------- | ------------------- |
| $$P(t)=A+tb$$                             | Main ray equation   |
| $$A=(x_0,y_0,z_0)$$                       | Origin point        |
| $$b=(d_x,d_y,d_z)$$                       | Direction vector    |
| $$t \in \mathbb{R}$$                      | Distance/step value |
| $$P(t)=(x_0+td_x,; y_0+td_y,; z_0+td_z)$$ | Expanded form       |

---

# Formula Description

| Symbol | Description              |
| ------ | ------------------------ |
| $P(t)$ | Point on the ray         |
| $A$    | Start position           |
| $b$    | Direction vector         |
| $t$    | Scalar multiplier        |
| $tb$   | Scaled direction         |
| $A+tb$ | Final point after moving |

---

# Purpose of Each Part in Code

| Code Part     | Purpose                  |
| ------------- | ------------------------ |
| `self.orig`   | Stores starting point    |
| `self.dir`    | Stores direction         |
| `origin()`    | Returns starting point   |
| `direction()` | Returns direction vector |
| `at(t)`       | Computes point on ray    |

---

# How `at(t)` Works

| Step            | Formula  | Meaning              |
| --------------- | -------- | -------------------- |
| Start           | $$A$$    | Starting point       |
| Scale direction | $$tb$$   | Move by distance $t$ |
| Add together    | $$A+tb$$ | New point            |

Example:

If

$$
A=(1,2,3)
$$

and

$$
b=(4,5,6)
$$

for

$$
t=2
$$

then:

$$
P(2)=A+2b
$$

$$
=(1,2,3)+2(4,5,6)
$$

$$
=(1,2,3)+(8,10,12)
$$

$$
=(9,12,15)
$$

---

# Visual Meaning

| $t$ value | Result             |
| --------- | ------------------ |
| $t=0$     | At origin          |
| $t=1$     | One direction step |
| $t=2$     | Two steps forward  |
| $t=-1$    | One step backward  |

---

# Why Ray Tracing Needs This

| Use             | Purpose                     |
| --------------- | --------------------------- |
| Camera rays     | Shoot rays from camera      |
| Object hit test | Check if ray touches object |
| Shadow rays     | Test light blocking         |
| Reflection rays | Bounce from surfaces        |
| Refraction rays | Pass through glass          |

---

Think of it simply:

$$
\text{Ray} = \text{Start Point} + \text{Movement}
$$

Like:

**"Start here, go this way."**
