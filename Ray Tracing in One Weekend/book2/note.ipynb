This is a roadmap of what **Book 2: *Ray Tracing: The Next Week*** adds on top of your current ray tracer. I’ll break it down in a simple “what it means + why it matters” way so you can connect it to your code.

---

## 1. Motion Blur

**What it is:** Objects look blurry when they move.

**In ray tracing:**
Instead of rendering at one instant, you sample time over an interval.

**Why it matters:**

* Makes animation look realistic (like spinning wheels, moving balls)
* Without it, motion looks “stiff” or fake

---

## 2. Bounding Volume Hierarchies (BVH)

**What it is:** A tree structure that groups objects into boxes.

**In ray tracing:**
Instead of checking ray vs *every object*, you:

* First check big bounding boxes
* Then go deeper only if needed

**Why it matters:**

* Huge speed boost
* Makes complex scenes (1000s of objects) feasible

---

## 3. Texture Maps

**What it is:** Images wrapped onto 3D objects.

**In ray tracing:**
You map (u, v) coordinates on a surface → sample color from an image.

**Why it matters:**

* Realistic materials (wood, stone, faces, etc.)
* Removes need for procedural coloring only

---

## 4. Perlin Noise

**What it is:** Smooth “random-looking” noise.

**In ray tracing:**
Used to generate natural patterns:

* clouds
* marble
* fire-like textures

**Why it matters:**

* Makes procedural textures look organic instead of flat noise

---

## 5. Quadrilaterals (Quads)

**What it is:** Flat 2D surfaces in 3D space.

**In ray tracing:**
Basic surface primitive besides spheres.

From quads you can build:

* triangles (mesh rendering foundation)
* disks
* rectangles
* complex models

**Why it matters:**

* Moves you toward real-world 3D models (not just spheres)

---

## 6. Lights

**What it is:** Explicit light sources.

**In ray tracing:**
You add:

* point lights
* area lights (important for soft shadows)

**Why it matters:**

* Enables shadows, highlights, realism
* Moves beyond “self-lit materials”

---

## 7. Transforms

**What it is:** Moving/rotating/scaling objects.

**In ray tracing:**
You apply matrices or math transforms to objects/rays.

**Why it matters:**

* Place objects anywhere in the scene
* Rotate and scale without rewriting geometry

---

## 8. Volumetric Rendering

**What it is:** Rendering *stuff inside space*, not just surfaces.

Examples:

* fog
* smoke
* clouds

**In ray tracing:**
Rays accumulate color while passing through a medium.

**Why it matters:**

* Adds atmosphere
* Makes scenes look cinematic

---

## Big Picture (What this book is really teaching)

This book upgrades your ray tracer from:

> “A system that renders spheres”

to:

> “A real rendering engine”

Because you now get:

* speed (BVH)
* realism (motion blur, lights, textures)
* complexity (quads, transforms)
* atmosphere (volumes)
* procedural realism (Perlin noise)
